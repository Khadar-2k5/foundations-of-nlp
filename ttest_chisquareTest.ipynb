```python
"""t-Test and Chi-Square test to check whether a given sequence of words is a
collocation or not."""
```

```python
import nltk
nltk.download("punkt_tab")
nltk.download("stopwords")

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.collocations import BigramCollocationFinder
from nltk.metrics import BigramAssocMeasures

text = "Natural language processing is an exciting field of artificial intelligence. Machine learning models are widely used in text classification and sentiment analysis."

text = text.lower()

import gensim

token = gensim.utils.simple_preprocess(text)

englishstop = stopwords.words("english")

tokens = []

for word in token:
    if word.lower() not in englishstop:
        tokens.append(word)

# Create bigram finder
finder = BigramCollocationFinder.from_words(tokens)


# Loop through all bigrams
for bigram, freq in finder.ngram_fd.items():

    t_score = finder.score_ngram(BigramAssocMeasures.student_t, *bigram)

    chi_sq = finder.score_ngram(BigramAssocMeasures.chi_sq, *bigram)

    # Check chi-square significance
    if chi_sq is not None and chi_sq > 8.32:
        colloc = "YES"
    else:
        colloc = "NO"

    print("Bigram:",bigram," : "," Frequency:",freq," T-Score: ",t_score," ChiSquare-Score: ",chi_sq," Collocation: ",colloc);
```
