```python

from nltk.tokenize import word_tokenize,sent_tokenize

text = "I LOVE INDIA! and worship the India Very much.NLP is interesting. It helps computers understand language."
r = word_tokenize(text)
r1=sent_tokenize(text)
print(r)
print(r1)

from nltk.corpus import stopwords
sw=set(stopwords.words("english"))
ft=[]
for x in r:
    if x.lower() not in sw:
        ft.append(x.lower())
print("Filtered Tokens:",ft)
                       
```

```python
from nltk.stem import PorterStemmer
p1=PorterStemmer()
afs=[]
for x in ft:
    afs.append(p1.stem(x))
print("AFTER PORTER STEMMER: ",afs)

from nltk.stem import WordNetLemmatizer
w=WordNetLemmatizer()
afw=[]
for x in afs:
    afw.append(w.lemmatize(x))
print("AFTER WORDNET LEMMATIZER: ",afw)
    
    
```

```python
from nltk import CFG,ChartParser
from nltk.tokenize import word_tokenize
text="i saw man with telescope"
t=word_tokenize(text)
g=CFG.fromstring("""
S -> NP VP
VP -> V NP | VP PP
PP -> P NP
V -> 'saw'
NP -> 'i' | 'man' | 'telescope'
P -> 'with'
""")
p=ChartParser(g)
for t in p.parse(t):
    print(t)
    t.draw()
```

```python
from nltk import CFG,ChartParser
from nltk.tokenize import word_tokenize
text="the red cat chases the dog"
x1=word_tokenize(text)
g=CFG.fromstring("""
S -> NP VP
    NP -> DT NN | DT JJ NN
    VP -> VBZ NP | VBZ
    DT -> 'the' | 'a'
    JJ -> 'red'
    NN -> 'cat' | 'dog'
    VBZ -> 'chases' | 'sleeps'
    """)
p=ChartParser(g)
for t in p.parse(x1):
    t.draw()
    t.pretty_print()
```

```python
from nltk import CFG,ChartParser
from nltk.tree import Tree
from nltk.tokenize import word_tokenize
text="the cat sat on the mat"
t=word_tokenize(text)
c=CFG.fromstring("""
S -> NP VP
  NP -> Det N
  VP -> V PP
  PP -> P NP
  Det -> 'The' | 'the'
  N -> 'cat' | 'mat'
  V -> 'sat'
  P -> 'on'
  """)
p=ChartParser(c)
y=list(p.parse(t))
for i,tree in enumerate(y):
    print(tree.pformat())
    tree.draw()
```
