```python
import pandas as pd
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize

texts = [
    "I love NLP",
    "Machine learning is fun",
    "AI is the future"
]

sentences = []
for text in texts:
    words = word_tokenize(text.lower())
    sentences.append(words)

model = Word2Vec(sentences,vector_size=50,window=3,min_count=1)

print("Vector of 'ai':")
print(model.wv['ai'][:5])
print("\nSimilarity (ai, learning):")
print(model.wv.similarity('ai', 'learning'))
print("\nWords similar to 'ai':")
print(model.wv.most_similar('ai'))
```
