```python
import numpy as np
states = ["CP", "IP"]
obs = ["lem", "icet", "cola"]
start_prob = [1, 0]
trans_prob = [
    [0.7, 0.3],
    [0.5, 0.5]
]
emit_prob = {
    "CP": {"lem": 0.3, "icet": 0.1, "cola": 0.6},
    "IP": {"lem": 0.2, "icet": 0.7, "cola": 0.1}
}
N = len(states)
T = len(obs)
delta = np.zeros((N, T))
psi = np.zeros((N, T), dtype=int)
for i in range(N):
    delta[i, 0] = start_prob[i] * emit_prob[states[i]][obs[0]]
print("Initial Delta:\n")
print(delta)
for t in range(1, T):
    for j in range(N):
        probs = []
        for i in range(N):
            probs.append(delta[i, t-1] * trans_prob[i][j])
        psi[j, t] = np.argmax(probs)
        delta[j, t] = np.max(probs) * emit_prob[states[j]][obs[t]]
    print("\nDelta at time", t)
    print(delta)
best_last = np.argmax(delta[:, T-1])
max_prob = delta[best_last, T-1]
path = [best_last]
for t in range(T-1, 0, -1):
    path.insert(0, psi[path[0], t])
best_states = []
for i in path:
    best_states.append(states[i])
print("\nMost likely state sequence:", best_states)
print("Probability:", max_prob)
```
