```python
from collections import Counter

train_data = [
    ("I went to the bank to deposit money","finance"),
    ("She withdrew cash from the bank","finance"),
    ("The river bank was full of trees","river"),
    ("Children played near the river bank","river")
]

test_sentences = [
    "He deposited money in the bank",
    "They sat near the bank of the river"
]

# SEPARATE WORDS BY SENSE
sense_words = {}

for sentence, sense in train_data:

    words = sentence.lower().split()

    if sense not in sense_words:
        sense_words[sense] = []

    sense_words[sense].extend(words)


word_counts = {}
total_words = {}

for s in sense_words:

    word_counts[s] = Counter(sense_words[s])

    total_words[s] = sum(word_counts[s].values())


for sentence in test_sentences:

    words = sentence.lower().split()

    scores = {}

    for s in word_counts:

        score = 1

        for w in words:

            score *= (word_counts[s][w] + 1) / (
                total_words[s] + len(word_counts[s])
            )

        scores[s] = score

    print("\nSentence:", sentence)
    print("Predicted sense:", max(scores, key=scores.get))
```
