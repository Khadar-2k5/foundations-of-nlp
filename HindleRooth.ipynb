```python
"""Hindle and Rooth algorithm for solving the attachment ambiguity"""
```

```python
import pandas as pd 
import math 
from nltk.tokenize import word_tokenize 
from nltk import bigrams 
 
# read corpus (works for csv or txt) 
text="The boy saw with a telescope.The girl saw with binoculars.The man ate with a spoon.The child played with toys.The boy saw with a camera."

text=text.lower()
# tokenize 
tokens = word_tokenize(text.lower()) 
# unigram counts 
word_counts = {} 
for w in tokens: 
    word_counts[w] = word_counts.get(w,0) + 1 
# bigram counts 
bigram_counts = {} 
for b in bigrams(tokens): 
    bigram_counts[b] = bigram_counts.get(b,0) + 1 
# input words 
noun = input("Enter noun: ").lower() 
verb = input("Enter verb: ").lower() 
prep = input("Enter preposition: ").lower() 
# counts 
n = word_counts.get(noun,0) 
v = word_counts.get(verb,0) 
pn = bigram_counts.get((noun,prep),0) 
pv = bigram_counts.get((verb,prep),0) 
# probabilities 
p_v = pv/(v+1e-10) 
p_n = pn/(n+1e-10) 
# lambda 
lam = math.log2((p_v+1e-10)/(p_n+1e-10)) 
print("\nLambda =",lam) 
if lam>0: 
    print("Attach to VERB") 
else: 
    print("Attach to NOUN")
```
