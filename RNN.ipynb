```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
eng = ["i am good","you are good","i am bad"]
fra = ["je suis bon","tu es bon","je suis mauvais"]

tok_eng = Tokenizer()
tok_fra = Tokenizer()
tok_eng.fit_on_texts(eng)
tok_fra.fit_on_texts(fra)
X = tok_eng.texts_to_sequences(eng)
X = pad_sequences(X)
y = tok_fra.texts_to_sequences(fra)
y = pad_sequences(y, maxlen=X.shape[1])
model = Sequential()
model.add(Embedding(len(tok_eng.word_index) + 1,8,input_length=X.shape[1]))

model.add(SimpleRNN(16, return_sequences=True))

model.add(Dense(len(tok_fra.word_index) + 1,activation='softmax'))

model.compile(optimizer='adam',loss='sparse_categorical_crossentropy')
model.fit(X, y, epochs=200, verbose=0)
test = "i am good"
test_seq = tok_eng.texts_to_sequences([test])
test_seq = pad_sequences(test_seq,maxlen=X.shape[1])

pred = model.predict(test_seq, verbose=0)

result = []
for step in pred[0]:
    index = np.argmax(step)
    word = tok_fra.index_word.get(index, "")
    if word != "":
        result.append(word)
print("Input:", test)
print("Translation:", " ".join(result))
```
