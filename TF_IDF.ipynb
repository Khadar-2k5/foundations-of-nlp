```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, TextVectorization
texts = [
    "I love this movie",
    "This film is amazing",
    "I hate this movie",
    "This film is terrible",
    "Wonderful acting and story",
    "Bad screenplay and boring scenes"
]
labels = np.array([1, 1, 0, 0, 1, 0])
split = int(0.8 * len(texts))
train_texts = texts[:split]
test_texts = texts[split:]
train_labels = labels[:split]
test_labels = labels[split:]

bow = TextVectorization(output_mode='count')
bow.adapt(train_texts)
X_train_bow = bow(train_texts)
X_test_bow = bow(test_texts)

model_bow = Sequential()
model_bow.add(Dense(8, input_dim=X_train_bow.shape[1], activation='relu'))
model_bow.add(Dense(1, activation='sigmoid'))
model_bow.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model_bow.fit(X_train_bow, train_labels, epochs=5, verbose=0)
loss_bow, acc_bow = model_bow.evaluate(
    X_test_bow,
    test_labels,
    verbose=0
)

print("BoW Accuracy:", acc_bow)
tfidf = TextVectorization(output_mode='tf-idf')
tfidf.adapt(train_texts)
X_train_tfidf = tfidf(train_texts)
X_test_tfidf = tfidf(test_texts)

model_tfidf = Sequential()
model_tfidf.add(Dense(8, input_dim=X_train_tfidf.shape[1], activation='relu'))
model_tfidf.add(Dense(1, activation='sigmoid'))
model_tfidf.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
model_tfidf.fit(
    X_train_tfidf,
    train_labels,
    epochs=5,
    verbose=0
)
loss_tfidf, acc_tfidf = model_tfidf.evaluate(
    X_test_tfidf,
    test_labels,
    verbose=0
)

print("TF-IDF Accuracy:", acc_tfidf)
test = ["I love this film"]
print("\nBoW Prediction:")
print(model_bow.predict(bow(test)))
print("\nTF-IDF Prediction:")
print(model_tfidf.predict(tfidf(test)))
```
